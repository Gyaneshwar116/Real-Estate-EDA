**OUTLIER DETECTION**

```python
Q1 = df["Total_Price"].quantile(0.25)
Q3 = df["Total_Price"].quantile(0.75)

IQR = Q3 - Q1

Lower = Q1 - 1.5 * IQR
Upper = Q3 + 1.5 * IQR

((df["Total_Price"] < Lower) | (df["Total_Price"] > Upper)).sum()
```

```python
df[(df["Total_Price"] < Lower) | (df["Total_Price"] > Upper)]
((df["Total_Price"] < Lower) | (df["Total_Price"] > Upper)).mean()*100
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.ticker import FuncFormatter
```

```python
plt.figure(figsize=(15,8))

plt.subplot(1,3,1)
sns.boxplot(y=df["Area"])
plt.title("Area")
plt.ylim(0, 10000)

plt.subplot(1,3,2)
sns.boxplot(y=df["Price_Per_Sqft"])
plt.title("Price_Per_Sqft")
plt.ylim(0, 50000)

plt.subplot(1,3,3)
sns.boxplot(y=df["Total_Price"])
plt.title("Total_Price")
plt.ylim(0, 200000000)
plt.ylabel("Total_Price (Crores)")
plt.gca().yaxis.set_major_formatter(FuncFormatter(lambda x, pos: f"{x/1e7:.1f} Cr"))

plt.tight_layout()
plt.show()
```

The IQR method detected around **10%** outliers in Total_Price, and
the boxplots also revealed outliers in Area, Price_Per_Sqft, and Total_Price.
Instead of removing these observations, I verified whether they
represented genuine market values or data entry errors.
The analysis indicated that **these records correspond to luxury villas and premium apartments located in high-end areas, making them valid business observations rather than anomalies**.
Since they are valid business observations, I retained them to
preserve the real distribution of the Hyderabad property market.

```python
num_cols = ["BHK","Area","Price_Per_Sqft","Total_Price","Current_Floor","Total_Floors"]
```

```python
plt.figure(figsize=(15,8))

for i, col in enumerate(num_cols,1):
    plt.subplot(3,2,i)
    sns.histplot(df[col], kde=True,bins=10)
    plt.title(col)

plt.tight_layout()
plt.show()
```

The histogram shows that 3 BHK is the most common property configuration in the dataset, followed by 2 BHK and 4 BHK. As the number of bedrooms increases beyond 4 BHK, the frequency of properties decreases significantly. The distribution is positively skewed, indicating that properties with 5 or more bedrooms are relatively rare. These larger properties likely belong to the premium or luxury housing segment, which caters to a smaller group of buyers.

The Area distribution is positively skewed, with most properties concentrated between approximately 1000 and 2500 sqft. As the property area increases, the number of available listings decreases significantly, resulting in a long right tail. This indicates that small and medium-sized residential properties dominate the Hyderabad market, while large-area luxury villas and premium homes represent a smaller segment of the available listings.

**They are concentrated on the left, and the tail extends to the right.**

The Total_Price distribution is positively skewed, with most properties concentrated in the affordable to mid-price range (approximately ₹50 lakhs to ₹1 crore). As the property price increases, the number of available listings decreases significantly, resulting in a long right tail. The tail represents premium and luxury properties, which are naturally fewer in number. Overall, the dataset indicates that the Hyderabad residential market is dominated by affordable and mid-priced property listings, while luxury homes form a smaller market segment.

```python
df[num_cols].skew()
```

All numerical variables exhibit positive (right) skewness, indicating that most properties are concentrated at lower values, while a relatively small number of premium or luxury properties extend the distribution toward higher values. Among all variables, Area (10.91), Total_Price (8.42), and Price_Per_Sqft (6.27) show extremely high positive skewness due to the presence of luxury properties with exceptionally large sizes and high prices. Current_Floor is also positively skewed, whereas Total_Floors (0.89) shows moderate positive skewness, indicating comparatively less variation.

```python
top10 = df.groupby("Location")["Total_Price"].mean().sort_values(ascending=False).head(10).index

pivot = df[df["Location"].isin(top10)].pivot_table(
    index="Location",
    columns="Property_Type",
    values="Total_Price",
    aggfunc="mean"
)

# Convert the pivot table to float to handle pd.NA values, converting them to np.nan
pivot = pivot.astype(float)

plt.figure(figsize=(8,6))
sns.heatmap(pivot, annot=True, fmt=".0f", cmap="coolwarm")
plt.title("Top 10 Locations by Average Property Price")
plt.show()
```
